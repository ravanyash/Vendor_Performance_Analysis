```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import sqlite3
import math
from scipy.stats import ttest_ind
import scipy.stats as stats
warnings.filterwarnings('ignore')
```

# Loading the dataset

```python
# creating database connection
conn = sqlite3.connect('inventory.db')

# fetching vendor summary data
df = pd.read_sql_query("SELECT * FROM vendor_sales_summary",conn)
df.head()
```

# Exploratory Data Analysis

```python
# summary stats
```

```python
df.describe().T
```

```python
# Distribution Plots for Numerical Columns
numerical_cols = df.select_dtypes(include=np.number).columns

n = len(numerical_cols)
cols = 4
rows = math.ceil(n / cols)

plt.figure(figsize=(5 * cols, 4 * rows))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(rows, cols, i)
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,10))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(rows, cols, i)
    sns.boxplot(y=df[col])
    plt.title(col)
    plt.tight_layout()

plt.show()
```

# Summary Statistics Insights
## 1. Negative & Zero Values
- Gross Profit: The minimum value of -52,002.78 indicates that certain products or transactions generated significant losses. This may occur when products are sold below their purchase cost or due to high operational expenses.
- Profit Margin: The minimum value is -∞ (negative infinity), which typically occurs when Sales Revenue is zero, causing a division-by-zero error during calculation. These records should be investigated before further analysis.
- Sales Quantity & Sales Revenue: Minimum values of 0 indicate that some products were purchased but never sold. These products may represent slow-moving, obsolete, or newly stocked inventory.
## 2. Presence of Outliers
- Purchase Price & Actual Price: The maximum values (5,681.81 and 7,499.99) are substantially higher than their respective mean values (24.39 and 35.64). This suggests the presence of premium or luxury products, which create high-value outliers.
- Freight Cost: Freight costs vary widely from 0.09 to 257,032.07, indicating significant differences in shipment sizes, transportation methods, or supplier locations. Extremely high freight costs should be reviewed for operational efficiency.
- Stock Turnover: Stock turnover ranges from 0 to 274.50. A value of 0 indicates products with no sales, while very high turnover values indicate products that sell rapidly. Values greater than 1 suggest that sales exceeded purchases during the analysis period, likely because sales were fulfilled using inventory carried over from previous purchase cycles.
## 3. Business Observations
- The dataset contains a mix of fast-moving and slow-moving inventory, indicating varying product demand.
- A small number of products contribute disproportionately to sales value, reflecting the presence of high-priced items.
- Negative profits and zero sales highlight products that may require pricing adjustments, promotional strategies, or inventory optimization.
- High variability in freight costs suggests opportunities to optimize logistics and supplier management.

# Positive Proftable vendor analysis

```python
query = """
SELECT *
FROM vendor_sales_summary
WHERE GrossProfit > 0
  AND ProfitMargin > 0
  AND TotalSalesQuantity > 0;
"""

filtered_df = pd.read_sql(query, conn)
```

```python
filtered_df
```

```python
filtered_df.describe().T
```

```python
# Distribution Plots for Numerical Columns
numerical_cols = filtered_df.select_dtypes(include=np.number).columns

n = len(numerical_cols)
cols = 4
rows = math.ceil(n / cols)

plt.figure(figsize=(5 * cols, 4 * rows))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(rows, cols, i)
    sns.histplot(filtered_df[col], kde=True, bins=30)
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# Distribution Plots for Numerical Columns
numerical_cols = filtered_df.select_dtypes(include=np.number).columns

n = len(numerical_cols)
cols = 4
rows = math.ceil(n / cols)

plt.figure(figsize=(18, 4 * rows))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(rows, cols, i)
    sns.boxplot(y=filtered_df[col])
    plt.title(col)
    plt.tight_layout()

plt.show()
```

```python
# Count Plots for Categorical Columns
categorical_cols = ["VendorName","Description"]

plt.figure(figsize=(12,5))
for i, col in enumerate(categorical_cols):
    plt.subplot(1, 2, i+1)
    sns.countplot(y=filtered_df[col], order=filtered_df[col].value_counts().index[:10]) # Top 10 categories
    plt.title(f"Count Plot for {col}")

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(12,8))
correaltion_matrix = filtered_df[numerical_cols].corr()
sns.heatmap(correaltion_matrix, annot = True, fmt=".2f" , cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

### Correlation Insights

- **Purchase Price vs Sales & Profit:** Purchase Price has a negligible negative correlation with **Total Sales Dollars (-0.01)** and **Gross Profit (-0.02)**, indicating that changes in purchase price have minimal impact on overall sales revenue and profitability.

- **Purchase Quantity vs Sales Quantity:** Total Purchase Quantity and Total Sales Quantity show an almost perfect positive correlation **(1.00)**, suggesting that inventory purchased is efficiently converted into sales, reflecting effective inventory management.

- **Profit Margin vs Total Sales Price:** Profit Margin has a weak negative correlation with Total Sales Price **(-0.18)**, indicating that higher selling prices do not necessarily translate into higher profit margins, possibly due to increased procurement costs or competitive pricing strategies.

- **Stock Turnover vs Profitability:** Stock Turnover has a negligible negative correlation with Gross Profit **(-0.04)** but a moderate positive correlation with Profit Margin **(0.40)**. This suggests that faster inventory turnover is associated with better profit margins but does not significantly increase total gross profit.

- **Purchase Price vs Actual Price:** Purchase Price and Actual Price exhibit an almost perfect positive correlation **(0.99)**, indicating strong consistency between procurement cost and recorded product price.

- **Gross Profit vs Total Sales Dollars:** Gross Profit has a very strong positive correlation with Total Sales Dollars **(0.98)**, showing that higher sales revenue directly contributes to increased profitability.

- **Gross Profit vs Total Purchase Dollars:** Gross Profit is strongly positively correlated with Total Purchase Dollars **(0.96)**, suggesting that vendors with larger purchase volumes tend to generate higher profits.

- **Total Sales Dollars vs Total Sales Price:** Total Sales Dollars and Total Sales Price have a strong positive correlation **(0.83)**, indicating that higher sales prices are associated with greater sales revenue.

- **Total Purchase Dollars vs Total Excise Tax:** Total Purchase Dollars and Total Excise Tax are strongly positively correlated **(0.88)**, showing that larger purchase volumes result in higher excise tax expenses.
